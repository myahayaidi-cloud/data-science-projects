## 1.1 Contexte
L'amélioration de l'état de santé des populations constitue l'un des principaux défis du développement durable, particulièrement en Afrique de l'Ouest où les maladies transmissibles et les complications liées à la santé maternelle continuent d'exercer une forte pression sur les systèmes de santé. Malgré les progrès enregistrés au cours des deux dernières décennies, plusieurs pays de la région demeurent confrontés à une incidence élevée du paludisme ainsi qu'à des niveaux encore préoccupants de mortalité maternelle.

L'exploitation des données ouvertes publiées par l'Organisation mondiale de la Santé (OMS) et Our World in Data offre aujourd'hui l'opportunité d'analyser l'évolution de ces indicateurs sanitaires sur une longue période et de mettre en évidence les disparités existantes entre les pays.
Dans ce contexte, la Data Analysis constitue un outil d'aide à la décision particulièrement pertinent. En transformant les données brutes en visualisations et indicateurs synthétiques, elle permet d'identifier les tendances, de comparer les performances des pays et de fournir aux décideurs des éléments objectifs pour orienter les politiques de santé.

## 4.2 Nettoyage du jeu de données sur le paludisme

Le fichier sur le paludisme présente déjà une structure relativement simple. Les opérations de nettoyage visent principalement à conserver les informations utiles pour l'étude.

Étape 1 : Sélection des variables utiles

Le fichier contient uniquement quatre variables, qui sont toutes pertinentes pour notre analyse.

```python
# Sélection des variables utiles

malaria = malaria[
    [
        "Entity",
        "Code",
        "Year",
        "Incidence of malaria (per 1,000 population at risk)"
    ]
]
```

Étape 2 : Renommer la variable principale

Le nom de la colonne est relativement long. Nous le simplifions afin de faciliter les manipulations ultérieures.

```python
malaria = malaria.rename(
    columns={
        "Incidence of malaria (per 1,000 population at risk)":
        "Malaria_Incidence"
    }
)
```

Étape 3 : Conserver uniquement les pays d'Afrique de l'Ouest

Notre étude porte exclusivement sur les seize pays de l'Afrique de l'Ouest.

```python
west_africa = [
    "Benin",
    "Burkina Faso",
    "Cape Verde",
    "Cote d'Ivoire",
    "Gambia",
    "Ghana",
    "Guinea",
    "Guinea-Bissau",
    "Liberia",
    "Mali",
    "Mauritania",
    "Niger",
    "Nigeria",
    "Senegal",
    "Sierra Leone",
    "Togo"
]

malaria_wa = malaria[
    malaria["Entity"].isin(west_africa)
].copy()
```

Étape 4 : Limiter la période d'étude

Même si le fichier couvre les années 2000 à 2024, notre étude s'arrête en 2020 afin d'être compatible avec les autres indicateurs.

```python
malaria_wa = malaria_wa[
    malaria_wa["Year"].between(2000, 2020)
]
```

Étape 5 : Vérification de la qualité

```python
print("Dimensions :", malaria_wa.shape)

print("\nValeurs manquantes")
print(malaria_wa.isna().sum())

print("\nDoublons :", malaria_wa.duplicated().sum())
```

À ce stade, le fichier du paludisme est prêt pour les analyses et pourra être fusionné avec les autres jeux de données.

## 4.3 Nettoyage du jeu de données sur la mortalité maternelle

Le processus est similaire.

Étape 1 : Sélection des variables utiles

```python
maternal = maternal[
    [
        "Entity",
        "Code",
        "Year",
        "Maternal mortality ratio"
    ]
]
```

Étape 2 : Renommer la variable

```python
maternal = maternal.rename(
    columns={
        "Maternal mortality ratio":
        "Maternal_Mortality"
    }
)
```

```python
maternal_wa = maternal[
    maternal["Entity"].isin(west_africa)
].copy()
```

Étape 5 : Contrôle qualité

## 4.4 Nettoyage du jeu de données UHC

C'est ici que le travail devient plus intéressant.

Contrairement aux deux premiers jeux de données, celui-ci contient : des régions OMS, des regroupements géographiques, des données mondiales, de nombreuses colonnes administratives. Il va falloir donc procéder méthodiquement.

Étape 1 : Conserver uniquement les pays

```python
uhc = uhc[
    uhc["SpatialDimension"] == "COUNTRY"
].copy()
```

Étape 2 : Renommer les colonnes

```python
uhc = uhc.rename(
    columns={
        "SpatialDimensionValueCode": "Code",
        "TimeDim": "Year",
        "NumericValue": "UHC_Coverage"
    }
)
```

Étape 3 : Ajouter le nom du pays

Le fichier UHC ne contient que le code ISO. Nous créons donc une table de correspondance entre les codes ISO et les noms des pays.

```python
country_map = {
    "BEN": "Benin",
    "BFA": "Burkina Faso",
    "CPV": "Cape Verde",
    "CIV": "Cote d'Ivoire",
    "GMB": "Gambia",
    "GHA": "Ghana",
    "GIN": "Guinea",
    "GNB": "Guinea-Bissau",
    "LBR": "Liberia",
    "MLI": "Mali",
    "MRT": "Mauritania",
    "NER": "Niger",
    "NGA": "Nigeria",
    "SEN": "Senegal",
    "SLE": "Sierra Leone",
    "TGO": "Togo"
}

uhc["Entity"] = uhc["Code"].map(country_map)
```

Étape 4 : Conserver uniquement les pays de l'étude

```python
uhc = uhc[
    uhc["Code"].isin(country_map.keys())
]
```

Étape 5 : Limiter la période

```python
uhc = uhc[
    uhc["Year"].between(2000, 2020)
]
```

Étape 6 : Conserver uniquement les variables utiles

```python
uhc = uhc[
    [
        "Entity",
        "Code",
        "Year",
        "UHC_Coverage"
    ]
]
```

Étape 7 : Vérification finale

```python
print(uhc.shape)

uhc.isna().sum()

uhc.duplicated().sum()
```

## 4.5 Comparaison des trois jeux de données nettoyés

### 6.2.3 Évolution temporelle des indicateurs

Comment les indicateurs évoluent-ils dans le temps ? Pour répondre à cette question, nous calculons d'abord la moyenne régionale annuelle.

```python
evolution = (
    health
    .groupby("Year")[
        [
            "Malaria_Incidence",
            "Maternal_Mortality",
            "UHC_Coverage"
        ]
    ]
    .mean()
)

evolution.head()
```

La courbe met en évidence une diminution progressive de l'incidence du paludisme dans les pays d'Afrique de l'Ouest étudiés entre 2000 et 2020. L'incidence moyenne passe d'environ 360 cas à près de 230 cas pour 1 000 habitants exposés au risque, soit une réduction importante sur l'ensemble de la période.

Une légère phase de stabilisation est observée autour des années 2008 à 2010, traduisant un ralentissement temporaire des progrès avant la reprise de la tendance à la baisse.

Cette évolution suggère une amélioration de la lutte contre le paludisme dans la région. Elle est cohérente avec le renforcement des stratégies de prévention et de prise en charge mises en œuvre au cours des deux dernières décennies, notamment la distribution de moustiquaires imprégnées d'insecticide, l'amélioration de l'accès aux traitements antipaludiques et les campagnes de sensibilisation.

```python
#Evolution temporelle de la mortalité maternelle

plt.figure(figsize=(10,5))

sns.lineplot(
    data=evolution,
    x=evolution.index,
    y="Maternal_Mortality",
    linewidth=2
)

plt.title("Evolution de la mortalité maternelle")

plt.show()
```

Ces écarts s’expliquent par des différences en matière de conditions environnementales, de niveau de développement socio-économique, mais également par l’efficacité des politiques nationales de lutte contre le paludisme, notamment la distribution des moustiquaires imprégnées, les campagnes de prévention et l’accès aux traitements.

### 6.3.4 Synthèse comparative des trois indicateurs
Cette section vise à réaliser une synthèse des performances sanitaires des seize pays d'Afrique de l'Ouest en comparant leurs rangs pour chacun des trois indicateurs étudiés. L'objectif est d'identifier les pays les plus performants, ceux présentant des performances intermédiaires ainsi que ceux confrontés à des difficultés simultanées en matière de lutte contre le paludisme, de santé maternelle et de couverture sanitaire universelle.

```python
#Étape 1 : Création des rangs

#Paludisme
malaria_rank["Rang_Paludisme"] = range(1, len(malaria_rank)+1)

malaria_rank.head()
```

```python
#Mortalité maternelle
maternal_rank["Rang_Maternal"] = range(1, len(maternal_rank)+1)

maternal_rank.head()
```

```python
#UHC
uhc_rank["Rang_UHC"] = range(1, len(uhc_rank)+1)

uhc_rank.head()
```

```python
#Étape 2 : Fusion des trois classements

comparison = (
    malaria_rank[["Entity","Rang_Paludisme","Malaria_Incidence"]]
    .merge(
        maternal_rank[["Entity","Rang_Maternal","Maternal_Mortality"]],
        on="Entity"
    )
    .merge(
        uhc_rank[["Entity","Rang_UHC","UHC_Coverage"]],
        on="Entity"
    )
)

comparison
```

```python
#Étape 3 : Classement global par rang moyen (Contrairement à l'ISCP, ici on ne crée pas un nouvel indicateur sanitaire. On calcule simplement la moyenne des positions obtenues dans les trois classements.)

comparison["Rang_Moyen"] = (
    comparison[
        ["Rang_Paludisme",
         "Rang_Maternal",
         "Rang_UHC"]
    ]
    .mean(axis=1)
)

comparison = comparison.sort_values("Rang_Moyen")

comparison.round(2)
```

```python
#Étape 4 : Identifier les profils sanitaires

def profil(r):

    if r <= 5:
        return "Très performant"

    elif r <= 10:
        return "Performance intermédiaire"

    else:
        return "Vulnérable"

comparison["Profil"] = comparison["Rang_Moyen"].apply(profil)

comparison
```

La comparaison des classements met en évidence une forte hétérogénéité des performances sanitaires en Afrique de l'Ouest. Le Cap-Vert, le Ghana et le Sénégal se distinguent comme les pays présentant les meilleurs résultats sur l'ensemble des indicateurs, tandis que des pays tels que la Guinée, la Sierra Leone, le Niger, le Benin et le Burkina Faso figurent parmi ceux confrontés aux plus grands défis sanitaires. Les autres tel que la Gambie occupent une position relativement favorable. 

Certains pays présentent des performances contrastées, avec de bons résultats pour un indicateur mais des difficultés persistantes pour les autres, ce qui souligne la complexité des systèmes de santé dans la région.

Pour l'incidence du paludisme et la mortalité maternelle, les valeurs ont été inversées de manière à ce qu'une valeur élevée corresponde toujours à une meilleure performance.

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

radar = uemoa_resume.copy()

radar[[
    "Paludisme",
    "Mortalité",
    "UHC"
]] = scaler.fit_transform(

    radar[
        [
            "Malaria_Incidence",
            "Maternal_Mortality",
            "UHC_Coverage"
        ]
    ]

)

# inversion des deux indicateurs où une faible valeur est préférable

radar["Paludisme"] = 1 - radar["Paludisme"]

radar["Mortalité"] = 1 - radar["Mortalité"]

radar
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Variables à représenter
categories = ["Paludisme", "Mortalité", "UHC"]
N = len(categories)

angles = np.linspace(0, 2*np.pi, N, endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(8,8))
ax = plt.subplot(111, polar=True)

# Tracer chaque pays
for i in range(len(radar)):
    values = radar.loc[i, ["Paludisme", "Mortalité", "UHC"]].tolist()
    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=radar.loc[i, "Entity"]
    )

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=12)

ax.set_ylim(0,1)

plt.title(
    "Profils sanitaires des pays de l'UEMOA",
    fontsize=15,
    weight="bold"
)

plt.legend(
    bbox_to_anchor=(1.30,1.05)
)

plt.show()
```

### 6.4.5.5 Classification des profils sanitaires par K-Means
Cette analyse applique l'algorithme K-Means aux trois indicateurs sanitaires normalisés afin d'identifier des groupes de pays présentant des profils similaires. Contrairement à un simple classement, cette approche permet de mettre en évidence des typologies de systèmes de santé au sein de l'UEMOA.

```python
#Préparation des données
from sklearn.cluster import KMeans

X = radar[["Paludisme", "Mortalité", "UHC"]]
```

```python
#Choix du nombre de clusters (Compte tenu du faible nombre de pays (8), 3 clusters est un bon compromis entre simplicité et capacité à distinguer les profils).
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

radar["Cluster"] = kmeans.fit_predict(X)
```

```python
#Résultat
radar[
    [
        "Entity",
        "Cluster",
        "Paludisme",
        "Mortalité",
        "UHC"
    ]
].sort_values("Cluster")
```

```python
#Visualisation des clusters (Nous utilisons ici une projection sur deux dimensions par ACP (PCA) pour représenter les groupes de façon lisible).
from sklearn.decomposition import PCA

pca = PCA(n_components=2)

coord = pca.fit_transform(X)

radar["PC1"] = coord[:,0]
radar["PC2"] = coord[:,1]

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=radar,
    x="PC1",
    y="PC2",
    hue="Cluster",
    s=150,
    palette="Set1"
)

for i in range(len(radar)):
    plt.text(
        radar.loc[i,"PC1"]+0.01,
        radar.loc[i,"PC2"]+0.01,
        radar.loc[i,"Entity"],
        fontsize=10
    )

plt.title("Classification des pays de l'UEMOA par K-Means")

plt.xlabel("Composante principale 1")

plt.ylabel("Composante principale 2")

plt.tight_layout()

plt.show()
```

| ------------- | --------------------------------------------------------- | -------------------- | -------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| **Cluster 0** | **Guinée-Bissau**                                         | Profil atypique      | La Guinée-Bissau constitue un cas particulier. Elle est isolée par l'algorithme en raison de la combinaison d'un indice UHC relativement élevé, d'une très forte mortalité maternelle et d'un niveau intermédiaire d'incidence du paludisme. Ce profil contraste avec celui des autres pays de l'UEMOA.        |
| **Cluster 1** | **Burkina Faso, Bénin, Côte d'Ivoire, Mali, Niger, Togo** | Profil intermédiaire | Ce groupe rassemble la majorité des pays de l'UEMOA. Ces pays font face à des défis sanitaires similaires.                               |

# CHAPITRE 6.6 Analyse statistique

### 6.6.1 Analyse des corrélations
Cette section vise à mesurer les relations statistiques entre les trois indicateurs retenus dans cette étude : l'incidence du paludisme, la mortalité maternelle et l'indice de couverture sanitaire universelle (UHC). L'objectif est d'évaluer si l'amélioration de la couverture sanitaire est associée à une diminution de l'incidence du paludisme et de la mortalité maternelle.

```python
#Corrélation de Pearson
corr = health[
    [
        "Malaria_Incidence",
        "Maternal_Mortality",
        "UHC_Coverage"
    ]
].corr(method="pearson")

corr
```

```python
#Heatmap
plt.figure(figsize=(7,6))

sns.heatmap(
    corr,
    annot=True,
    cmap="RdBu_r",
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=.5
)

plt.title(
    "Matrice de corrélation de Pearson",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.show()
```

| Relation | Coefficient de corrélation (r) | Interprétation |
| --- | --- | --- |
| **Paludisme – Mortalité maternelle** | **+0.29** | Corrélation positive faible : les pays où le paludisme est plus fréquent tendent à présenter une mortalité maternelle légèrement plus élevée. Cela traduit une vulnérabilité sanitaire commune, mais pas une dépendance forte. |
| **Paludisme – UHC** | **−0.60** | Corrélation négative modérée : une meilleure couverture sanitaire est associée à une baisse de l’incidence du paludisme. Ce lien suggère que l’accès aux soins et à la prévention joue un rôle déterminant dans la réduction du risque. |
| **Mortalité maternelle – UHC** | **−0.51** | Corrélation négative moyenne : plus la couverture sanitaire est élevée, plus la mortalité maternelle diminue. Ce résultat confirme l’impact direct des politiques de santé universelle sur la survie des mères. |

Les corrélations négatives entre UHC et les deux autres indicateurs montrent que la qualité du système de santé est un facteur transversal de performance sanitaire.

La corrélation positive entre paludisme et mortalité maternelle traduit une cohérence épidémiologique : les pays les plus exposés aux maladies infectieuses sont aussi ceux où les conditions obstétricales restent précaires.

Aucun coefficient ne dépasse ±0.7, ce qui indique des relations modérées, non déterministes — chaque indicateur conserve sa spécificité.

```python
#Corrélation de Spearman
corr_s = health[
    [
        "Malaria_Incidence",
        "Maternal_Mortality",
        "UHC_Coverage"
    ]
].corr(method="spearman")

corr_s
```

```python
# Création de la heatmap de Spearman
plt.figure(figsize=(6, 5))
sns.heatmap(
    corr_s,
    annot=True,          # Affiche les coefficients
    cmap="RdBu_r",       # Palette de couleurs rouge-bleu inversée
    center=0,            # Centre la palette autour de zéro
    vmin=-1, vmax=1,     # Échelle de corrélation
    linewidths=0.5,      # Séparation entre les cases
    cbar_kws={"shrink": 0.8, "label": "Coefficient de corrélation"}  # Barre de couleur
)

plt.title("Matrice de corrélation de Spearman", fontsize=14, fontweight="bold")
plt.show()
```

### 6.6.3 Régressions linéaires
Cette section vise à évaluer quantitativement l'association entre la couverture sanitaire universelle (UHC) et les deux principaux indicateurs de santé étudiés : l'incidence du paludisme et la mortalité maternelle.

Deux modèles de régression linéaire simple sont estimés :

- Modèle 1 : Effet de l'indice UHC sur l'incidence du paludisme.
- Modèle 2 : Effet de l'indice UHC sur la mortalité maternelle.

Les résultats seront interprétés à partir des coefficients de régression, du coefficient de détermination (R²) et des tests de significativité (p-value).

```python
#Modèle 1 : UHC → Paludisme
import statsmodels.api as sm

# Variable explicative
X = health["UHC_Coverage"]

# Ajout de la constante
X = sm.add_constant(X)

# Variable dépendante
Y = health["Malaria_Incidence"]

# Estimation
model_paludisme = sm.OLS(Y, X).fit()

# Résultats
print(model_paludisme.summary())
```

```python
#Modèle 2 : UHC → Mortalité maternelle
X = health["UHC_Coverage"]

X = sm.add_constant(X)

Y = health["Maternal_Mortality"]

model_mortalite = sm.OLS(Y, X).fit()

print(model_mortalite.summary())
```

### 6.6.4 Tests statistiques
Cette section évalue la significativité statistique des relations entre les trois indicateurs étudiés à l'aide des coefficients de corrélation de Pearson et de Spearman.

Le coefficient de Pearson mesure l'intensité d'une relation linéaire entre deux variables quantitatives, tandis que le coefficient de Spearman mesure l'intensité d'une relation monotone basée sur les rangs. Les deux approches sont utilisées afin de vérifier la robustesse des résultats obtenus.

Les hypothèses testées sont les suivantes :

- H₀ : il n'existe pas de relation statistiquement significative entre les deux variables.
- H₁ : il existe une relation statistiquement significative entre les deux variables.

Le seuil de significativité retenu est de 5 % (α = 0,05).

```python
#Pearson
from scipy.stats import pearsonr

variables = [
    ("Malaria_Incidence", "Maternal_Mortality"),
    ("Malaria_Incidence", "UHC_Coverage"),
    ("Maternal_Mortality", "UHC_Coverage")
]

resultats_pearson = []

for x, y in variables:

    r, p = pearsonr(health[x], health[y])

    resultats_pearson.append({
        "Variables": f"{x} ↔ {y}",
        "Coefficient r": r,
        "p-value": p
    })

pearson_df = pd.DataFrame(resultats_pearson)

pearson_df.round(4)
```

```python
#Spearman
from scipy.stats import spearmanr

resultats_spearman = []

for x, y in variables:

    rho, p = spearmanr(health[x], health[y])

    resultats_spearman.append({
        "Variables": f"{x} ↔ {y}",
        "Coefficient ρ": rho,
        "p-value": p
    })

spearman_df = pd.DataFrame(resultats_spearman)

spearman_df.round(4)
```

```python
#Tableau comparatif
comparaison = pd.DataFrame({

    "Variables": pearson_df["Variables"],

    "Pearson (r)": pearson_df["Coefficient r"],

    "Pearson p-value": pearson_df["p-value"],

    "Spearman (ρ)": spearman_df["Coefficient ρ"],

    "Spearman p-value": spearman_df["p-value"]

})

comparaison.round(4)
```

##### Interprétation des coefficients de corrélation

Les coefficients de corrélation de Pearson et de Spearman mettent en évidence des relations statistiquement significatives entre les trois indicateurs étudiés. Les p-values inférieures à 0,001 pour l'ensemble des tests permettent de rejeter l'hypothèse nulle d'absence de corrélation, ce qui confirme que les relations observées sont statistiquement significatives au seuil de 5 %.

La relation entre l'incidence du paludisme et la mortalité maternelle est positive mais faible selon le coefficient de Pearson (r = 0,29) et très faible selon le coefficient de Spearman (ρ = 0,18). Cela signifie que les pays présentant une incidence plus élevée du paludisme tendent également à enregistrer une mortalité maternelle plus importante, mais cette association demeure limitée. Le paludisme apparaît ainsi comme un facteur associé à la santé maternelle, sans constituer à lui seul un déterminant majeur de la mortalité maternelle.

La relation entre l'indice de couverture sanitaire universelle (UHC) et l'incidence du paludisme est négative et forte selon Pearson (r = -0,60) et également forte selon Spearman (ρ = -0,62). Cette forte concordance entre les deux coefficients montre qu'une meilleure couverture sanitaire est systématiquement associée à une diminution de l'incidence du paludisme, indépendamment de la méthode de corrélation utilisée. Ce résultat confirme les observations issues des analyses descriptives et des modèles de régression.

La relation entre l'indice UHC et la mortalité maternelle est négative et modérée selon Pearson (r = -0,51) et modérée selon Spearman (ρ = -0,41). Ces résultats indiquent qu'une amélioration de la couverture sanitaire universelle est associée à une réduction de la mortalité maternelle. Bien que cette relation soit moins forte que celle observée avec le paludisme, elle demeure statistiquement significative et met en évidence le rôle important de la couverture sanitaire dans l'amélioration de la santé maternelle.

Dans l'ensemble, les analyses de corrélation confirment les résultats obtenus par les modèles de régression. Elles montrent que la couverture sanitaire universelle constitue l'indicateur le plus fortement associé aux deux principaux indicateurs de santé étudiés, avec une association particulièrement marquée pour l'incidence du paludisme. À l'inverse, la relation directe entre le paludisme et la mortalité maternelle est plus faible, suggérant que ces deux phénomènes sont influencés par des déterminants multiples.

##### Conclusion de la section 6.6

Les analyses statistiques réalisées à travers les corrélations, les régressions linéaires et les tests de significativité convergent vers une même conclusion : une meilleure couverture sanitaire universelle est systématiquement associée à une diminution de l'incidence du paludisme et de la mortalité maternelle pour le groupe des pays. Cette association est particulièrement forte pour le paludisme, pour lequel près de 36 % de la variabilité est expliquée par l'indice UHC.

Ces résultats sont cohérents avec les travaux de l'Organisation mondiale de la Santé, qui considère la couverture sanitaire universelle comme un levier essentiel pour améliorer les indicateurs de santé, notamment grâce à un meilleur accès aux services de prévention, de diagnostic et de prise en charge.

### 6.7.2 Implications pour les politiques de santé

Les résultats obtenus soulignent l'importance de poursuivre les investissements dans les systèmes de santé afin d'améliorer la couverture des services essentiels. Le renforcement des infrastructures sanitaires, la disponibilité du personnel qualifié et l'amélioration de l'accès aux soins constituent des leviers majeurs pour réduire simultanément l'incidence du paludisme et la mortalité maternelle.

Les différences observées entre les pays suggèrent également que les bonnes pratiques mises en œuvre dans les pays les plus performants pourraient inspirer les stratégies nationales des pays confrontés à des difficultés persistantes.

Les résultats plaident enfin pour une approche intégrée des politiques de santé, dans laquelle la lutte contre le paludisme, la santé maternelle et le renforcement de la couverture sanitaire universelle sont considérés comme des priorités complémentaires plutôt que comme des interventions indépendantes.

### 6.7.3 Limites de l'étude

Cette étude présente plusieurs limites qu'il convient de prendre en considération.

Premièrement, elle repose exclusivement sur des données secondaires provenant de bases internationales. Bien que ces données soient reconnues pour leur fiabilité, elles peuvent comporter des différences de qualité entre les pays et au fil du temps.

Deuxièmement, l'analyse s'appuie sur trois indicateurs seulement : l'incidence du paludisme, la mortalité maternelle et l'indice de couverture sanitaire universelle. D'autres facteurs susceptibles d'influencer ces indicateurs, tels que les dépenses de santé, le niveau de pauvreté, le niveau d'éducation, la densité du personnel de santé ou les conditions climatiques, n'ont pas été intégrés.

Troisièmement, les modèles de régression utilisés mettent en évidence des associations statistiques mais ne permettent pas d'établir des relations de causalité entre les variables étudiées.

Enfin, la période d'étude est limitée aux années 2000 à 2020. Les évolutions plus récentes, notamment les effets de la pandémie de COVID-19 sur les systèmes de santé, ne sont pas prises en compte.

### 6.7.4 Recommandations et perspectives de recherche

Au regard des résultats obtenus, plusieurs recommandations peuvent être formulées :

poursuivre le renforcement de la couverture sanitaire universelle afin d'améliorer l'accès équitable aux services de santé essentiels ;
renforcer les programmes de prévention et de prise en charge du paludisme, notamment dans les pays où l'incidence demeure élevée ;
améliorer les services de santé maternelle en développant les soins prénatals, l'assistance qualifiée à l'accouchement et les soins obstétricaux d'urgence ;
accroître les investissements dans les infrastructures sanitaires et la formation des professionnels de santé ;
renforcer les systèmes d'information sanitaire afin de disposer de données plus complètes et plus récentes pour le suivi des politiques publiques.

Pour les recherches futures, il serait pertinent d'intégrer d'autres variables explicatives, telles que les dépenses publiques de santé, le produit intérieur brut par habitant, le niveau d'éducation, la pauvreté, la densité du personnel de santé ou encore les facteurs environnementaux. L'utilisation de modèles économétriques plus avancés, tels que les modèles de données de panel, permettrait également d'approfondir l'analyse des déterminants des performances sanitaires.
